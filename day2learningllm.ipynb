```python
# We created an agent and attached a tool (calculator function) to it.
# This allows the agent to perform mathematical calculations when needed.

# Inside the function, the text written in triple quotes (""" ... """)
# is called a *docstring*.

# A docstring is used to describe what the function does.
# In LangChain, this docstring is very important because:
# 👉 The agent reads this description to understand when to use the tool.

# For example, if the docstring says:
# "Evaluate a mathematical expression",
# then the agent will use this tool whenever it sees a math-related query.

# Function structure:
# - Function name: calculate → name of the tool
# - Parameter: expression → input given to the function
# - Docstring: explains the purpose of the tool

# So overall:
# The docstring acts like instructions for the agent,
# helping it decide when and how to use the calculator tool.
```

```python
import os
from groq import Groq

client = Groq(api_key=os.environ["GROQ_API_KEY"])

chat = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role": "user", "content": "Hello"}],
)
print(chat.choices[0].message.content)
```
