```python
#using search agent tool to search for information on the internet.

from agents import WebSearchTool


INSTRUCTIONS = """
You are a helpful assistant that can search the web for information.    
When you receive a query, use the web search tool to find relevant information and provide a concise answer.
""" 
search_agent = Agent(
    name="SearchAgent", 
    instructions=INSTRUCTIONS,
    model="gpt-4.1-mini",
    tools=[WebSearchTool(search_context_size="low")],  
)
```

```python
#without trace we can run the agent to search for information on the internet.
message = "what is pydantic please explain and also provide your source of information ?"

response = await Runner.run(search_agent, message)
print(response) 
```
